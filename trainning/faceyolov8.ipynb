```python
from ultralytics import YOLO

# Load mô hình YOLOv12
model = YOLO("yolov8n.pt")  # Hoặc dùng YOLOv8 nếu YOLOv12 chưa có

# Train model với dataset của bạn
model.train(data="/kaggle/input/facedetec/data.yaml", epochs=100, imgsz=640, batch=16, device="cuda")
```

```python
from ultralytics import YOLO

# Load model đã được train (hoặc model best.pt đã lưu)
model = YOLO("/kaggle/working/runs/detect/train/weights/best.pt")  # hoặc model cuối epoch: last.pt

# Chạy đánh giá trên tập validation đã cấu hình trong data.yaml
metrics = model.val()

# In các metrics
print(metrics)
```

```python
results = model.val()
metrics = results.box

# Truy cập đúng các chỉ số: KHÔNG có ngoặc tròn
print(f"Precision: {metrics.mp:.3f}")
print(f"Recall: {metrics.mr:.3f}")
print(f"mAP@0.5: {metrics.map50:.3f}")
print(f"mAP@0.5:0.95: {metrics.map:.3f}")
```
